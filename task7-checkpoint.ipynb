```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import PowerTransformer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score,confusion_matrix,precision_score,recall_score,f1_score,classification_report)
from mlxtend.plotting import plot_confusion_matrix
from sklearn.linear_model import RidgeClassifier , Lasso
```

```python
df = load_breast_cancer(as_frame=True).frame
df.columns = df.columns.str.replace(" ","_")
```

```python
#1. Load the dataset using pandas.
#2. Display first 5 rows.
#3. Check dataset info (columns, datatypes).
#4. Check missing values.
#5. Statistical summary using describe().
#6. Visualize data (histograms, correlation heatmap).
#df.head()
#df.info()
#df.isnull().sum()
#df.describe()
plt.figure(figsize=(20,10))
#sns.heatmap(df.corr(),annot = True)
num = df.select_dtypes("float64")
for i,col in enumerate(num):
    plt.subplot(5,6,i+1)
    sns.histplot(df[col],kde=True)
```

```python
#1. Handle missing values (fill or drop).
#2. Encode categorical variables (Label Encoding or One-Hot Encoding).
#3. Feature scaling (StandardScaler or MinMaxScaler).
#4. Split dataset into train and test sets (80/20).
x = df.drop("target",axis=1)
y = df.target
x_train , x_test  , y_train , y_test  =train_test_split(x  ,  y ,   test_size  =0.2)
```

```python
pt = PowerTransformer(method='yeo-johnson')
scaler = StandardScaler()

#train
transformedDataTrain = pt.fit_transform(x_train) 
x_train_scaled = scaler.fit_transform(transformedDataTrain)

#test
transformedDataTest = pt.transform(x_test) 
x_test_scaled = scaler.transform(transformedDataTest)
```

```python
#1. Import LogisticRegression from sklearn.
#2. Train the model using training data.
#3. Predict on test data.

model = LogisticRegression(max_iter = 200)
model.fit(x_train_scaled , y_train)
preds = model.predict(x_test_scaled)
#4. Explain how Logistic Regression works briefly (Sigmoid Function).
#it split the data into 2 classes 0 and 1 so it is perfect to take decision so Logistic Regression is linear Regression with sigmoid
```

```python
#1. Accuracy Score.
#2. Confusion Matrix.
#3. Precision, Recall, F1-score.
#4. Plot confusion matrix.

accuracy = accuracy_score(y_test, preds)
print("Accuracy:", accuracy)

cm = confusion_matrix(y_test, preds)
print("Confusion Matrix:\n", cm)

precision = precision_score(y_test, preds)
print("Precision:", precision)

recall = recall_score(y_test, preds)
print("Recall:", recall)

f1 = f1_score(y_test, preds)
print("F1 Score:", f1)

#5. Explain results (Is model good or not?)
#the result are good because accuarcy is more than 95% of random data which is perfect
```
