```python
# Insights and Observations

## 1. Dataset Overview
- The dataset contains 45,211 records.
- Most features are numerical, with some categorical variables (job, marital, education, contact, poutcome).
- The target variable is likely imbalanced (based on yes/no distribution in plots).
## 2. Missing Values Analysis
- The dataset has missing values mainly in categorical features:
  - job → 0.63%
  - education → 4.10%
  - contact → 28.79%
  - poutcome → 81.74% (very high missing rate)

👉 This indicates that:
- poutcome is heavily incomplete and may need to be dropped or carefully treated.
- contact also has a large amount of missing data and may affect model performance.
## 3. Numerical Features Insights
From statistical summary:

- Age:
  - Average age is ~41 years.
  - Most customers are between 33 and 48 years.

- Balance:
  - Highly skewed with extreme values (min = -8019, max = 102127).
  - Indicates presence of strong outliers.

- Duration:
  - Highly variable (0 to 4918 seconds).
  - Strong right skew observed in histogram.

- Campaign:
  - Most clients were contacted 1–3 times.
  - Some extreme cases up to 63 contacts.

- pdays:
  - Many values are -1, indicating no previous contact.

- previous:
  - Most values are 0 → very few previous interactions.
## 4. Correlation Insights
- Strongest positive correlation:
  - pdays and previous (0.45) → indicates relationship between past interactions.

- Weak correlations between most numerical features and each other.
- No strong linear relationship between features → model may benefit from non-linear algorithms.
## 5. Distribution Insights (Histograms)
- age → slightly right-skewed distribution.
- balance → highly right-skewed with extreme outliers.
- duration → strongly right-skewed.
- campaign, pdays, previous → heavily concentrated at low values.
## 6. Categorical Features Insights (from plots)
- Marital status:
  - Married clients are the most common group.

- Education:
  - Secondary education dominates.

- Housing loan:
  - Majority have housing loans.

- Personal loan:
  - Most clients do not have personal loans.

- Contact method:
  - Cellular is the dominant contact type.

- Month:
  - Most contacts occur in May, followed by other months unevenly.

- Default:
  - Almost all clients have no credit default.
## 7. Key Observations
- The dataset is highly imbalanced in several features.
- Strong skewness exists in multiple numerical columns.
- Missing data is significant in poutcome and contact.
- Many outliers exist in balance, duration, and campaign.
- Previous marketing interactions are mostly zero.
```

```python
#fill the nan value in poutcome with nonexisting because itis important eature which cannot be deleted
df["poutcome"] = df["poutcome"].fillna("nonexisting")
#fill the nan value in contact with the same two value in it but randomly
df["contact"] = df["contact"].apply(lambda x: np.random.choice(['cellular', 'telephone']) if pd.isna(x) else x).astype("object")
#because the low of nan value we drop this row
df.dropna(subset=["education", "job"], inplace=True)
#drop the column duration cause it makes the the model bais
df = df.drop("duration",axis=1)
#make the -1 in seperate column to prevent confuse
df["was_not_previously_contacted"] =  (df["pdays"] == -1).astype(int)
df["pdays"] = df["pdays"].replace(-1, 0)
```

```python
#to create an model we cannot treatment with -1 to prevent the model confiusion and helping in scaling
#so we create a new feature wich is if the pdays == -1 the new feature "was_not_previously_contacted" =1
#then we replace the -1 in pdays column to 0
df["total_contacts"] = df["campaign"] + df["previous"]
```

```python
#see the skew in data to know what if  distribution is balanced or shifted to one side so itis need to transformation
for i in df.select_dtypes("number"):
    print(f"{i} = {df[i].skew()}")
```

```python
for i in df.select_dtypes("object"):
  print(f"'{i}'",end=",")
```

```python
num = ['age','balance','day','campaign','pdays','previous','was_not_previously_contacted','total_contacts']
cat = ['job','marital','education','default','housing','loan','contact','month','poutcome']
for i in num:
    pt = PowerTransformer(method='yeo-johnson')
    df[i] = pt.fit_transform(df[[i]])
```

```python
for i in num:
    scaler = RobustScaler()
    df[i] = scaler.fit_transform(df[[i]])
```

```python
for i in cat:
    encoder = OneHotEncoder()
    encoded_data = encoder.fit_transform(df[[i]]).toarray()
    encoded_data = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(), index=df.index)
    df = pd.concat([df, encoded_data], axis=1)
    df.drop(columns=[i], inplace=True)
```

```python
df.columns
```

```python
df["y"] = df["y"].map({"yes": 1, "no": 0})
```

```python
x=df.drop(["y"],axis=1)
y=df.y
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE()
x_train_res, y_train_res = smote.fit_resample(x_train, y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier

print("RandomForestClassifier")
print("n_estimators,max_depth,min_samples_split,train_acc,test_acc,f1")

for n_estimators in [50, 100, 200, 300]:
    for max_depth in [5, 10, None, 15]:
        for min_samples_split in [2, 5, 10, 12]:

            model = RandomForestClassifier(
                n_estimators=n_estimators,
                max_depth=max_depth,
                min_samples_split=min_samples_split,
            )

            model.fit(x_train_res, y_train_res)

            preds = model.predict(x_test)
            train_preds = model.predict(x_train_res)
            train_acc = accuracy_score(y_train_res, train_preds)
            test_acc = accuracy_score(y_test, preds)
            f1 = f1_score(y_test, preds)

            print(n_estimators, max_depth, min_samples_split,
                  train_acc*100, test_acc*100, f1*100)
```

```python
from sklearn.svm import SVC

print("SVM")
print("C,kernel,train_acc,test_acc,f1")

for C in [0.1, 1, 10, 100]:
    for kernel in ["linear", "rbf", "poly"]:

        model = SVC(C=C, kernel=kernel)

        model.fit(x_train_res, y_train_res)

        preds = model.predict(x_test)
        train_preds = model.predict(x_train_res)

        train_acc = accuracy_score(y_train_res, train_preds)
        test_acc = accuracy_score(y_test, preds)
        f1 = f1_score(y_test, preds)

        print(C, kernel,
              train_acc*100, test_acc*100, f1*100)
```

```python
from sklearn.ensemble import GradientBoostingClassifier

print("GradientBoostingClassifier")
print("n_estimators,learning_rate,max_depth,train_acc,test_acc,f1")

for n_estimators in [50, 100, 200]:
    for learning_rate in [0.01, 0.1, 0.2]:
        for max_depth in [3, 5, 7]:

            model = GradientBoostingClassifier(
                n_estimators=n_estimators,
                learning_rate=learning_rate,
                max_depth=max_depth
            )

            model.fit(x_train_res, y_train_res)

            preds = model.predict(x_test)
            train_preds = model.predict(x_train_res)

            train_acc = accuracy_score(y_train_res, train_preds)
            test_acc = accuracy_score(y_test, preds)
            f1 = f1_score(y_test, preds)

            print(n_estimators, learning_rate, max_depth,
                  train_acc*100, test_acc*100, f1*100)
```

```python
from xgboost import XGBClassifier

print("XGBoost")

for n_estimators in [50, 100, 200]:
    for lr in [0.01, 0.1, 0.2]:
        for depth in [3, 5, 7]:

            model = XGBClassifier(
                n_estimators=n_estimators,
                learning_rate=lr,
                max_depth=depth,
                eval_metric="logloss"
            )

            model.fit(x_train_res, y_train_res)
            preds = model.predict(x_test)
            f1 = f1_score(y_test, preds)
            train_preds = model.predict(x_train_res)
            train_acc = accuracy_score(y_train_res, train_preds)
            test_acc = accuracy_score(y_test, preds)

            print(n_estimators, lr, depth,train_acc*100 ,100*test_acc, f1*100)
```

```python
from sklearn.metrics import classification_report, precision_score, recall_score, f1_score

print("Precision:", precision_score(y_test, preds))
print("Recall:", recall_score(y_test, preds))
print("F1-score:", f1_score(y_test, preds))
print("\nClassification Report:\n")
print(classification_report(y_test, preds))
```

```python
from sklearn.metrics import roc_curve, auc

y_prob = model.predict_proba(x_test)

fpr1, tpr1, _ = roc_curve(y_test, y_prob[:, 1])
auc1 = auc(fpr1, tpr1)

fpr0, tpr0, _ = roc_curve(1 - y_test, y_prob[:, 0])
auc0 = auc(fpr0, tpr0)

plt.plot(fpr1, tpr1, label=f"Class 1 AUC = {auc1:.2f}")
plt.plot(fpr0, tpr0, label=f"Class 0 AUC = {auc0:.2f}", linestyle="--")

plt.plot([0, 1], [0, 1], "k--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve (Both Classes)")
plt.legend()
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV
results=[]
models = {
    "RandomForest": (
        RandomForestClassifier(),
        {
            'n_estimators': [50, 100, 200,300],
            'max_depth': [5, 10, None, 15],
            'min_samples_split': [2, 5, 10, 12]
        }
    ),

    "SVM": (
        SVC(),
        {

            'C': [0.1, 1, 10, 100],
            'kernel':["linear", "rbf", "poly"]
        }
    ),

    "GradientBoosting": (
        GradientBoostingClassifier(),
        {
            'n_estimators': [50, 100,200],
            'learning_rate': [0.01, 0.1,0.2],
            'max_depth': [3, 5,7]
        }
    ),

    "XGBoost": (
        XGBClassifier(eval_metric="logloss"),
        {
            'n_estimators': [50, 100,200],
            'learning_rate': [0.01, 0.1,0.2],
            'max_depth': [3, 5,7]
        }
    )
}

for name, (model, params) in models.items():

    print(f"\n{name}")
    print("-" * 40)

    grid = GridSearchCV(
        estimator=model,
        param_grid=params,
        scoring='f1',
        cv=5,
        n_jobs=-1
    )

    grid.fit(x_train_res, y_train_res)

    print("Best Parameters:", grid.best_params_)
    print("Best CV F1:", grid.best_score_)

    y_pred = grid.best_estimator_.predict(x_test)

    test_f1 = f1_score(y_test, y_pred)
    test_acc = accuracy_score(y_test, y_pred)

    results.append({
        "Model": name,
        "Best_CV_F1": grid.best_score_,
        "Test_F1": test_f1,
        "Test_Accuracy": test_acc
    })


    print("Test Accuracy:", test_acc * 100)
    print("Test F1:", test_f1 * 100)
```

```python
#the best model after multiple test on model this is the best cause the heighest f1 score which is 32% and acc test 88%
model5 = GradientBoostingClassifier(
    learning_rate=0.1,
    max_depth=7,
    n_estimators=50,
    random_state=42
)
model5.fit(x_train_res, y_train_res)
preds = model5.predict(x_test)
train_preds = model5.predict(x_train_res)
f1 = f1_score(y_test, preds)
train_acc = accuracy_score(y_train_res, train_preds)
test_acc = accuracy_score(y_test, preds)

print(train_acc*100 ,100*test_acc, f1*100)
```

#Limitations

Although the model achieved the best performance among all tested models, there are still several limitations:

Low F1-Score
The F1-score of 32% indicates that the model still struggles to correctly classify the minority class.
Possible Class Imbalance
The dataset may contain significantly more samples from one class than another, affecting model performance.
Overfitting Risk
Gradient Boosting models can overfit training data, especially with deeper trees (max_depth=7).
Limited Generalization
The model performance may decrease when applied to new unseen real-world data.
Feature Dependence
The model’s predictions depend heavily on the quality and relevance of input features.
Need for Further Optimization
Additional hyperparameter tuning, feature engineering, or advanced balancing techniques could improve results further.

```python
import xgboost as xgb
from sklearn.metrics import f1_score
import numpy as np

# Use raw imbalanced training data
neg = np.sum(y_train == 0)
pos = np.sum(y_train == 1)
ratio = neg / pos
print(f"Class ratio: {ratio:.1f}:1")

# XGBoost handles imbalance natively via scale_pos_weight
xgb_model = xgb.XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=ratio,   # the key parameter
    eval_metric='logloss',
    random_state=42
)

xgb_model.fit(x_train, y_train)   # <-- raw x_train, NOT x_train_res

# Find best threshold
probs = xgb_model.predict_proba(x_test)[:, 1]
results = {}
for t in np.arange(0.20, 0.70, 0.01):
    preds = (probs >= t).astype(int)
    results[t] = f1_score(y_test, preds)

best_t = max(results, key=results.get)
print(f"XGBoost (raw data) — Best threshold: {best_t:.2f} → F1: {results[best_t]:.4f}")
print(f"Baseline GradientBoosting F1: 0.4719")
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

# -----------------------------
# 1. Model (correct imbalance handling)
# -----------------------------
model = XGBClassifier(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=7.7,   # IMPORTANT for imbalance
    eval_metric="logloss",
    random_state=42
)

# -----------------------------
# 2. Train model
# -----------------------------
model.fit(x_train, y_train)

# -----------------------------
# 3. Get probabilities (IMPORTANT)
# -----------------------------
train_proba = model.predict_proba(x_train)[:, 1]
test_proba = model.predict_proba(x_test)[:, 1]

# -----------------------------
# 4. Threshold tuning (use your best or search)
# -----------------------------
threshold = 0.58  # you already found this earlier

train_pred = (train_proba >= threshold).astype(int)
test_pred = (test_proba >= threshold).astype(int)

# -----------------------------
# 5. Metrics
# -----------------------------
print("Train Accuracy:", accuracy_score(y_train, train_pred))
print("Test Accuracy:", accuracy_score(y_test, test_pred))

print("Train F1:", f1_score(y_train, train_pred))
print("Test F1:", f1_score(y_test, test_pred))
```

```python
from sklearn.metrics import f1_score
import numpy as np

probs = model.predict_proba(x_test)[:, 1]

thresholds = np.arange(0.1, 0.9, 0.01)

best_t = 0
best_f1 = 0

for t in thresholds:
    preds = (probs >= t).astype(int)
    score = f1_score(y_test, preds)
    if score > best_f1:
        best_f1 = score
        best_t = t

print("Best threshold:", best_t)
print("Best F1:", best_f1)
```
