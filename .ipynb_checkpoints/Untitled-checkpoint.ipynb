```python
from ucimlrepo import fetch_ucirepo 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
from imblearn.over_sampling import SMOTE
```

```python

# fetch dataset 
diabetes_130_us_hospitals_for_years_1999_2008 = fetch_ucirepo(id=296) 
  
# data (as pandas dataframes) 
X = diabetes_130_us_hospitals_for_years_1999_2008.data.features 
Y = diabetes_130_us_hospitals_for_years_1999_2008.data.targets 

```

```python
df = pd.concat([X,Y],axis=1)

df
```

```python
df.isnull().sum()
```

```python
df.isnull().mean()*100
```

```python
# Drop features with a very high percentage of missing values and features with limited relevance to the prediction task and that which has a static value(one value)
df = df.drop(["weight","max_glu_serum","A1Cresult","payer_code","medical_specialty","examide","citoglipton"],axis=1)
df["race"] = df["race"].fillna("Unknown")
df = df.dropna(subset=["diag_1", "diag_2", "diag_3"])
df.drop(df[df["gender"] == 'Unknown/Invalid'].index, inplace=True)
```

```python
#replace age range with a number 
age_mapping = {
    '[0-10)': 0, '[10-20)': 1, '[20-30)': 2, '[30-40)': 3,
    '[40-50)': 4, '[50-60)': 5, '[60-70)': 6, '[70-80)': 7,
    '[80-90)': 8, '[90-100)': 9
}
df['age'] = df['age'].map(age_mapping)
```

```python
categorical_cols = ["race","gender","age","admission_type_id","discharge_disposition_id","admission_source_id","number_diagnoses","change","diabetesMed","readmitted"]
plt.figure(figsize=(30,20))
for i,j in enumerate(categorical_cols):
    plt.subplot(5,2,i+1)
    sns.countplot(x=j,data = df)
```

```python
numeric_cols = ["time_in_hospital","num_lab_procedures","num_procedures","num_medications","number_outpatient","number_emergency","number_inpatient","number_diagnoses"]
plt.figure(figsize=(30,20))
for i,j in enumerate(numeric_cols):
    plt.subplot(4,2,i+1)
    sns.histplot(df[j],kde=True, edgecolor='black')
```

```python
numeric_cols = ["time_in_hospital","num_lab_procedures","num_procedures","num_medications","number_outpatient","number_emergency","number_inpatient","number_diagnoses"]
plt.figure(figsize=(30,20))
for i,j in enumerate(numeric_cols):
    plt.subplot(4,2,i+1)
    sns.boxplot(data=df,x="readmitted",y=j)
    
```

```python
z=['acetohexamide','tolbutamide','troglitazone',
              'glipizide-metformin','glimepiride-pioglitazone',
              'metformin-rosiglitazone','metformin-pioglitazone','miglitol','tolazamide']
for i in z:
    print(i)
    print(df[i].value_counts())
    print("-"*50)
```

```python
#drop it cause it has a little variation
second_line_drugs = ['nateglinide', 'chlorpropamide', 'acarbose', 'glyburide-metformin']

df['num_second_line_meds'] = (df[second_line_drugs] == 'Steady').sum(axis=1)

df['on_second_line_med'] = (df['num_second_line_meds'] > 0).astype(int)

df = df.drop(['acetohexamide','tolbutamide','troglitazone',
              'glipizide-metformin','glimepiride-pioglitazone',
              'metformin-rosiglitazone','metformin-pioglitazone','miglitol','tolazamide'],axis=1)

df = df.drop(columns=second_line_drugs)
```

```python
def map_icd9(code):
    try:
        code = float(code)
    except (ValueError, TypeError):
        return 'Other'
    if 390 <= code <= 459 or code == 785:
        return 'Circulatory'
    elif 460 <= code <= 519 or code == 786:
        return 'Respiratory'
    elif 520 <= code <= 579 or code == 787:
        return 'Digestive'
    elif 250 <= code < 251:
        return 'Diabetes'
    elif 800 <= code <= 999:
        return 'Injury'
    elif 710 <= code <= 739:
        return 'Musculoskeletal'
    elif 580 <= code <= 629 or code == 788:
        return 'Genitourinary'
    elif 140 <= code <= 239:
        return 'Neoplasms'
    else:
        return 'Other'

for col in ['diag_1', 'diag_2', 'diag_3']:
    df[col] = df[col].apply(map_icd9)
```

```python
df.iloc[:, -30:]
```

```python
onehot_encoders = {}

cat = ["race",'diag_1', 'diag_2', 'diag_3']
for i in cat:
    encoder = OneHotEncoder()
    encoded_data = encoder.fit_transform(df[[i]]).toarray()
    encoded_df = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(), index=df.index)
    df = pd.concat([df, encoded_df], axis=1)
    df.drop(columns=[i], inplace=True)
    onehot_encoders[i] = encoder
```

```python
label_encoders = {}
labelencoder = ["gender","change","diabetesMed"]
for i in labelencoder:
    encoder = LabelEncoder()
    df[i] = encoder.fit_transform(df[i])
    label_encoders[i] = encoder

```

```python
DRUG_ORDER = ['No', 'Down', 'Steady', 'Up']

ordinal_encoders = {}
ordinalencoder = ["metformin","repaglinide","glimepiride","glipizide","glyburide","pioglitazone","rosiglitazone","insulin"]

for i in ordinalencoder:
    encoder = OrdinalEncoder(categories=[DRUG_ORDER], handle_unknown='use_encoded_value', unknown_value=-1)
    df[[i]] = encoder.fit_transform(df[[i]])
    ordinal_encoders[i] = encoder
```

```python
df['readmitted'] = df['readmitted'].map({'<30': 1, '>30': 0, 'NO': 0})
```

```python
x=df.drop(["readmitted"],axis=1)
y=df.readmitted
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.preprocessing import RobustScaler
import numpy as np

heavy_skew_cols = ['number_outpatient', 'number_emergency', 'number_inpatient',
                    'repaglinide', 'num_second_line_meds']

moderate_cols = ['age', 'admission_type_id', 'discharge_disposition_id',
                  'admission_source_id', 'time_in_hospital', 'num_lab_procedures',
                  'num_procedures', 'num_medications', 'number_diagnoses',
                  'metformin', 'glimepiride', 'glipizide', 'glyburide',
                  'pioglitazone', 'rosiglitazone', 'insulin'] + heavy_skew_cols


for col in heavy_skew_cols:
    x_train[col] = np.log1p(x_train[col])
    x_test[col] = np.log1p(x_test[col])


scaler = RobustScaler()
x_train[moderate_cols] = scaler.fit_transform(x_train[moderate_cols])   
x_test[moderate_cols] = scaler.transform(x_test[moderate_cols])         
```

```python
from sklearn.model_selection import GridSearchCV

rf = RandomForestClassifier(max_features = 'sqrt', max_samples = 0.8, class_weight = "balanced")

rf_params = {"max_depth": [5, 7, 10],
             "n_estimators": range(100, 250, 30)}

rf_cv_model = GridSearchCV(rf, rf_params, cv=2, n_jobs=-1, scoring = 'roc_auc', verbose=2).fit(x_train, y_train)
print(rf_cv_model.best_score_)
print(rf_cv_model.best_params_)
```

```python
rf_model = rf_cv_model.best_estimator_
predicted_prob = rf_model.predict_proba(x_test)[:, 1]
predicted = rf_model.predict(x_test)
AUC_RF = roc_auc_score(y_test, predicted_prob)
print('AUC_RF = ', AUC_RF)
f1_pos = f1_score(y_test, predicted)
f1_neg = f1_score([0 if x == 1 else 1 for x in y_test], [0 if x == 1 else 1 for x in predicted])
print('F1_POS = ', f1_pos, 'F1_NEG = ', f1_neg)
HARMONIC_F1_RF = ((2 * f1_pos * f1_neg) / (f1_neg + f1_pos))
print('Harmonic_F1_RF =', HARMONIC_F1_RF)
```
